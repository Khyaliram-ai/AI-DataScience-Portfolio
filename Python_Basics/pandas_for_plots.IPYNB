```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('Toyota.csv', sep=',',na_values=['??', "????"],index_col=0 )
```

```python
df.dropna(axis = 0, inplace = True)
```

```python
plt.scatter(df['Age'], df['Price'],color='green')
plt.title('Scatter plot of Age vs Price of the cars')
plt.xlabel('Age (Months)')
plt.ylabel('Price (Euros)')
plt.show()
```

```python
plt.hist(df['KM'], color='green', edgecolor='white', bins=5)
plt.title('Histogram of KM')
plt.xlabel('Kilometers')
plt.ylabel('Frequency')
plt.show()
```

```python
fuelType = df['FuelType'].value_counts().index
counts = df['FuelType'].value_counts().values
index = np.arange(len(fuelType))

plt.bar(index, counts, color=['red', 'blue', 'cyan'])
plt.title('Bar plot of Fuel Types')
plt.xlabel('FuelTypes')
plt.ylabel('Frequency')
plt.xticks(index, fuelType, rotation=90)
plt.show()
```

```python
import seaborn as sns
df = pd.read_csv('Toyota.csv', index_col=0, na_values=['??', '????'])
```

```python
df.dropna(axis=0,  inplace=True)
```

```python
sns.set(style="darkgrid")
sns.regplot(x=df["Age"], y=df["Price"])
```

```python
sns.set(style="darkgrid")
sns.regplot(x=df["Age"], y=df["Price"],marker='*' ,fit_reg=False)
```

```python
sns.lmplot(x="Age", y="Price" ,data=df , fit_reg=False ,
           hue='FuelType',legend=True, palette='Set1')
```

```python
sns.displot(df['Age'],kde=False,bins =5)
```

```python
sns.countplot(x='FuelType',data=df,hue ="Automatic")
```
