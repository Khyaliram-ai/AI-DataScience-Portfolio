```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
department_salary =df.groupby("Department")["Salary"].mean()
department_salary.plot(kind="bar")

plt.title("Avarege Salary by Department")
plt.xlabel("Department")
plt.ylabel("Avarage Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["Age"],bins = 5)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```

```python
plt.scatter(df["Experience"],df["Salary"])
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

```python
plt.figure(figsize=(5, 5))
plt.pie(df["Age"] , labels=df["Name"] , autopct="%1.1f%%" , startangle=90)
plt.title("Age Distribution")
plt.tight_layout()
plt.show()
```

```python
sns.boxplot(data=df, x="Department", y="Salary")

plt.title("Salary Distribution by Department")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
