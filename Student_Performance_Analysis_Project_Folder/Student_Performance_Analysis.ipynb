## 4. Correlation Analysis

```python
df[["G1", "G2", "G3"]].corr()
```

```python
plt.imshow(df[["G1", "G2", "G3"]].corr())

plt.xticks(range(3), ["G1", "G2", "G3"])
plt.yticks(range(3), ["G1", "G2", "G3"])

plt.title("Correlation Between Grades")
plt.colorbar()
plt.show()
```

```python
plt.scatter(df["G1"], df["G3"])

plt.title("First Period Grade vs Final Grade")
plt.xlabel("G1")
plt.ylabel("G3")
plt.show()
```

```python
plt.scatter(df["G2"], df["G3"])

plt.title("Second Period Grade vs Final Grade")
plt.xlabel("G2")
plt.ylabel("G3")
plt.show()
```

```python
df["G3"].value_counts().sort_index()
```

```python
df["G3"].value_counts().sort_index().plot(kind="bar")

plt.title("Distribution of Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")
plt.show()
```

```python
df["result"] = df["G3"].apply(lambda x: "Pass" if x >= 10 else "Fail")

df["result"].value_counts()
```

```python
df["result"].value_counts().plot(kind="bar")

plt.title("Pass vs Fail Students")
plt.xlabel("Result")
plt.ylabel("Number of Students")
plt.show()
```

```python
print("Total Students:", len(df))
print("Average Final Grade:", df["G3"].mean())
print("Highest Final Grade:", df["G3"].max())
print("Lowest Final Grade:", df["G3"].min())
print("Average Absences:", df["absences"].mean())
```

## Key Findings

- The dataset contains 395 students.
- The average final grade (G3) is approximately 10.42.
- The highest final grade is 20, while the lowest is 0.
- The average number of absences is approximately 5.71.
- G1 and G3 show a strong positive correlation of approximately 0.80.
- G2 and G3 show a very strong positive correlation of approximately 0.90.
- Overall the dataset indicates that students with higher previous-period grades tend to have higher final grades.
